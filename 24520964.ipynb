```python
import numpy as np
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
```

```python
X, y = make_regression(n_samples=500, n_features=10, noise=10.0, bias=5.5, random_state=42)
```

Solve LR with sklearn (fit_intercept = True)

```python
model_true = LinearRegression(fit_intercept=True)
model_true.fit(X, y)
# Ghép intercept với coef
coef_sklearn_true = np.concatenate(([model_true.intercept_], model_true.coef_))

print(f"1. Sklearn (fit_intercept=True) : {coef_sklearn_true}")
```

Solve LR with sklearn (fit_intercept = False)

```python
X_b = np.c_[np.ones((X.shape[0], 1)), X]

model_false = LinearRegression(fit_intercept=False)
model_false.fit(X_b, y) 
coef_sklearn_false = model_false.coef_

print(f"2. Sklearn (fit_intercept=False): {coef_sklearn_false}")
```

Solve LR with normal equation

```python
beta_normal_eq = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)

print(f"3. Normal Equation              : {beta_normal_eq}")
```

Solve LR with gradient descent

```python
learning_rate = 0.05
n_iterations = 10000
m = len(X_b) 

np.random.seed(42)
beta_gd = np.random.randn(11) 

for iteration in range(n_iterations):
   
    gradients = 2/m * X_b.T.dot(X_b.dot(beta_gd) - y)

    beta_gd = beta_gd - learning_rate * gradients

print(f"4. Gradient Descent             : {beta_gd}")
```
